## Get the Data

```python
from pathlib import Path
import pandas as pd
import zipfile
import urllib.request
from sklearn.preprocessing import LabelEncoder
import numpy as np
```

```python
def load_data():
    zip_path = Path("datasets/portugal_data.zip")
    if not zip_path.is_file():
        Path("datasets").mkdir(parents=True, exist_ok=True)
        url = "https://github.com/Mankz111/datasets/raw/main/portugal_data.zip"
        urllib.request.urlretrieve(url, zip_path)
    with zipfile.ZipFile(zip_path, 'r') as portugal_dataset:
        portugal_dataset.extractall(path="datasets")
    return pd.read_csv(Path("datasets/portugal_listinigs.csv"), low_memory = False)
                
```

```python
data = load_data()
data.describe()
```

```python
missing_percentage = (data.isnull().sum() / len(data)) * 100
```

```python
cols_to_drop = ['GrossArea', 'Floor', 'PublishDate', 'ElectricCarsCharging', 'LotSize', 'BuiltArea', 'LivingArea', 'HasParking']
```

```python
new_data = data.drop(cols_to_drop, axis=1)
new_data.info()
```

```python
encoder = LabelEncoder()
```

```python
data_conserv = new_data["ConservationStatus"]
data_conserv_encoded = encoder.fit_transform(data_conserv)
```

```python
data_conserv_encoded
```

```python
print(encoder.classes_)
```

```python
mapping_conserv = {
    "Good condition" : 4,
    "Like new": 5,
    'New' : 6,
    'Reasonable' : 3,
    'Used': 2,
    'Needs renovation': 1,
    np.nan: 0  
}
```

```python
new_data["data_conserv_encoded"] = new_data["ConservationStatus"].map(mapping_conserv).fillna(0).astype(int)
```
